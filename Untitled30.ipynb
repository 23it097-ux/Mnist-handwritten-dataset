```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms
import numpy as np
import matplotlib.pyplot as plt
from opacus import PrivacyEngine
import warnings
warnings.filterwarnings('ignore')

# Set random seed
torch.manual_seed(42)
np.random.seed(42)

# Device setup (CPU as requested by instructions)
device = torch.device('cpu')
print(f"Using device: {device}")
```

```python
# Load MNIST dataset
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

train_dataset_full = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root='./data', train=False, download=True, transform=transform)

# Create a subset of 20,000 images for training as per instructions
train_subset_indices = list(range(20000))
train_subset = Subset(train_dataset_full, train_subset_indices)

train_loader = DataLoader(train_subset, batch_size=256, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=256, shuffle=False)

print(f"Train subset size: {len(train_subset)}")
print(f"Test set size: {len(test_dataset)}")
```

```python
class GroupNormCNN(nn.Module):
    def __init__(self):
        super(GroupNormCNN, self).__init__()
        # Standard CNN structure using GroupNorm instead of BatchNorm as required by Opacus
        self.feature_extractor = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=3, padding=1),
            nn.GroupNorm(4, 16), # 4 groups
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.GroupNorm(8, 32), # 8 groups
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 7 * 7, 128),
            nn.ReLU(),
            nn.Linear(128, 10)
        )

    def forward(self, x):
        x = self.feature_extractor(x)
        x = self.classifier(x)
        return x
```

```python
def train(model, train_loader, optimizer, device):
    model.train()
    criterion = nn.CrossEntropyLoss()
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

def evaluate(model, test_loader, device):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    return 100.0 * correct / total
```

```python
# Task 1: Train non-private baseline model
torch.manual_seed(42)
baseline_model = GroupNormCNN().to(device)
baseline_optimizer = optim.SGD(baseline_model.parameters(), lr=0.1, momentum=0.9)

print("Training non-private baseline model...")
for epoch in range(1, 4):
    train(baseline_model, train_loader, baseline_optimizer, device)
    acc = evaluate(baseline_model, test_loader, device)
    print(f"Epoch {epoch} | Test Accuracy: {acc:.2f}%")

baseline_acc = evaluate(baseline_model, test_loader, device)
print(f"Baseline (no DP) Final Test Accuracy: {baseline_acc:.2f}%")
```

```python
# Task 2: Train DP models for target epsilon = 1, 5, 10 using Opacus
target_epsilons = [10.0, 5.0, 1.0]
dp_results = {}
dp_models = {}

# Helper function to perform DP training using make_private_with_epsilon
def train_dp_model(target_epsilon):
    torch.manual_seed(42)
    model = GroupNormCNN().to(device)
    optimizer = optim.SGD(model.parameters(), lr=0.1, momentum=0.9)

    privacy_engine = PrivacyEngine()

    # make_private_with_epsilon automates finding the required noise_multiplier
    model, optimizer, dp_loader = privacy_engine.make_private_with_epsilon(
        module=model,
        optimizer=optimizer,
        data_loader=train_loader,
        target_epsilon=target_epsilon,
        target_delta=1e-5,
        epochs=3,
        max_grad_norm=1.0,
    )

    for epoch in range(1, 4):
        model.train()
        criterion = nn.CrossEntropyLoss()
        for images, labels in dp_loader:
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

    final_acc = evaluate(model, test_loader, device)
    epsilon_spent = privacy_engine.get_epsilon(delta=1e-5)
    return model, final_acc, epsilon_spent

for eps in target_epsilons:
    print(f"\nTraining DP Model with Target Epsilon = {eps}...")
    model_trained, test_acc, eps_spent = train_dp_model(eps)
    dp_results[eps] = {'test_acc': test_acc, 'eps_spent': eps_spent}
    dp_models[eps] = model_trained
    print(f"Target Epsilon: {eps} | Achieved Epsilon: {eps_spent:.2f} | Test Acc: {test_acc:.2f}%")
```

```python
# Task 3: Train one DP model with fixed noise_multiplier = 1.1
torch.manual_seed(42)
fixed_noise_model = GroupNormCNN().to(device)
fixed_noise_optimizer = optim.SGD(fixed_noise_model.parameters(), lr=0.1, momentum=0.9)

privacy_engine_fixed = PrivacyEngine()
fixed_noise_model, fixed_noise_optimizer, fixed_noise_loader = privacy_engine_fixed.make_private(
    module=fixed_noise_model,
    optimizer=fixed_noise_optimizer,
    data_loader=train_loader,
    noise_multiplier=1.1,
    max_grad_norm=1.0,
)

epoch_epsilons = []
print("\nTraining DP model with fixed noise_multiplier = 1.1...")
for epoch in range(1, 4):
    fixed_noise_model.train()
    criterion = nn.CrossEntropyLoss()
    for images, labels in fixed_noise_loader:
        images, labels = images.to(device), labels.to(device)
        fixed_noise_optimizer.zero_grad()
        outputs = fixed_noise_model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        fixed_noise_optimizer.step()

    eps_spent = privacy_engine_fixed.get_epsilon(delta=1e-5)
    epoch_epsilons.append(eps_spent)
    print(f"Epoch {epoch} | Epsilon Spent: {eps_spent:.2f}")
```

```python
# Task 4: Membership Inference Check (Mean confidence on train subset vs test)
def calculate_mean_max_softmax(model, data_loader, device):
    model.eval()
    confidences = []
    with torch.no_grad():
        for images, labels in data_loader:
            images = images.to(device)
            outputs = model(images)
            probabilities = torch.softmax(outputs, dim=1)
            max_probs, _ = torch.max(probabilities, dim=1)
            confidences.extend(max_probs.cpu().tolist())
    return np.mean(confidences)

# Prepare loaders for a matching subset of training data to speed up evaluation
eval_train_loader = DataLoader(train_subset, batch_size=256, shuffle=False)

print("Calculating Membership Inference metrics...")
baseline_train_conf = calculate_mean_max_softmax(baseline_model, eval_train_loader, device)
baseline_test_conf = calculate_mean_max_softmax(baseline_model, test_loader, device)
baseline_gap = baseline_train_conf - baseline_test_conf

dp_1_train_conf = calculate_mean_max_softmax(dp_models[1.0], eval_train_loader, device)
dp_1_test_conf = calculate_mean_max_softmax(dp_models[1.0], test_loader, device)
dp_1_gap = dp_1_train_conf - dp_1_test_conf
```

```python
# Task 5: Plotting
# Plot A: Accuracy vs Epsilon
eps_list = [1.0, 5.0, 10.0]
acc_list = [dp_results[eps]['test_acc'] for eps in eps_list]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(eps_list, acc_list, marker='o', color='b', label='DP Models')
plt.axhline(y=baseline_acc, color='r', linestyle='--', label=f'Baseline ({baseline_acc:.1f}%)')
plt.title('Accuracy vs Epsilon (Privacy-Utility Curve)')
plt.xlabel('Epsilon (ε)')
plt.ylabel('Test Accuracy (%)')
plt.grid(True)
plt.legend()

# Plot B: Epsilon vs Epoch
plt.subplot(1, 2, 2)
plt.plot([1, 2, 3], epoch_epsilons, marker='s', color='g', label='noise_multiplier = 1.1')
plt.title('Epsilon vs Epoch (Privacy Growth)')
plt.xlabel('Epoch')
plt.ylabel('Epsilon spent (ε)')
plt.xticks([1, 2, 3])
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.savefig('accuracy_vs_epsilon.png')
plt.show()
```
